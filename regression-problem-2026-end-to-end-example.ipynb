## LOADING LIBRARIES

```python
import pandas as pd
import matplotlib.pyplot as plt

import sklearn
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
import numpy as np
import warnings
warnings.filterwarnings("ignore")

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

## LOAD DATASET

```python
from pathlib import Path

# Resolve the dataset path robustly in this workspace
candidate_paths = [
    Path("student_module_pass_chance_regression.csv"),
    Path("dataset") / "student_module_pass_chance_regression.csv",
    Path("dataset") / "dataset" / "student_module_pass_chance_regression.csv",
]

file = next((str(p) for p in candidate_paths if p.exists()), str(candidate_paths[0]))
df = pd.read_csv(file)
df.head()
```

```python
encode = LabelEncoder()
df['Gender'] = encode.fit_transform(df['Gender'])
```

```python
df.head()
```

```python
df = df.drop(['Student_ID'], axis=1)
```

```python
df.isna().sum()
```

## DETECT OUTLIERS USING IQR

```python
# Select numerical columns
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

# Detect outliers using IQR
for column in numerical_columns:
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    
    print(f"{column}: {len(outliers)} outliers")
```

## HANDLE MISSING VALUES USING MEDIAN BCZ ALL FEATURES CONTAIN OUTLIERS

```python
df[numerical_columns] = df[numerical_columns].fillna(
    df[numerical_columns].median()
)
```

```python
df.isna().sum()
```

## REMOVE ALL DETECTED OUTLIERS

```python

print("Dataset before removing outliers:", df.shape)

# Remove outliers using IQR
for column in numerical_columns:
    
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    df = df[
        (df[column] >= lower_bound) &
        (df[column] <= upper_bound)
    ]

print("Dataset after removing outliers:", df.shape)
df.reset_index(drop=True, inplace=True)

df.head()
```

## Feature splitting

```python
X = df.drop(['Pass_Chance_Percentage'], axis=1)
y = df['Pass_Chance_Percentage']
print(X)
print(y)
```

## SPLIT DATASET

```python
X_train, X_test, y_train,y_test=train_test_split(X, y, test_size=.2, random_state=42)
X_train.shape
```

```python
X_test.shape
```

## Scale our data to prevent data leakage during model testing

```python
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_train
```

```python
X_test
```

## Model selection

## LINEAR REGRESSION

```python
model_lr = LinearRegression()
model_lr.fit(X_train_scaled, y_train)
```

```python
y_lr_pred = model_lr.predict(X_test_scaled)
# y_lr_pred
```

```python
y_test
```

```python
print("Linear Regression Results:")
print(f"R^2 Score: {r2_score(y_test, y_lr_pred):.4f}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_lr_pred):.4f}")
print(f"Mean Squared Error: {mean_squared_error(y_test, y_lr_pred):.4f}")
print(f"Root Mean Squared Error: {np.sqrt(mean_squared_error(y_test, y_lr_pred)):.4f}")
```

```python
print("Coefficients:", model_lr.coef_)
```

```python
print("Intercept:", model_lr.intercept_)
```
